simple statical imputation
mean imputation

```python
from sklearn.impute import SimpleImputer
import pandas as pd
df_mean = students.copy()
df_mean = df_mean.rename(columns={'Msrks': 'Marks', 'Attendence': 'Attendance'})
mean_imputer = SimpleImputer(strategy='mean')
df_mean[['Attendance', 'Marks']] = mean_imputer.fit_transform(
    df_mean[['Attendance', 'Marks']]
)

print(df_mean)
```

```python
from sklearn.impute import SimpleImputer
import pandas as pd
df_median = students.copy()
df_median = df_median.rename(columns={'Msrks': 'Marks', 'Attendence': 'Attendance'})
median_imputer = SimpleImputer(strategy='median')
df_median[['Attendance', 'Marks']] = median_imputer.fit_transform(
    df_median[['Attendance', 'Marks']]
)
print(df_median)
```

most frequend value imputation

```python
from sklearn.impute import SimpleImputer
import pandas as pd
df_const = students.copy()
df_const = df_const.rename(columns={'Msrks': 'Marks', 'Attendence': 'Attendance'})
num_imputer = SimpleImputer(strategy='constant', fill_value=0)
df_const[['Marks', 'Attendance']] = num_imputer.fit_transform(df_const[['Marks', 'Attendance']])
cat_imputer = SimpleImputer(strategy='constant', fill_value='Missing')
df_const[['Name']] = cat_imputer.fit_transform(df_const[['Name']])
print(df_const)
```

```python
from sklearn.impute import SimpleImputer
const_imputer = SimpleImputer(strategy="constant", fill_value="missing")
df_const = students.copy()
df_const[["Name"]] = const_imputer.fit_transform(df_const[["Name"]])

print(df_const)
```

```python
from sklearn.impute import KNNImputer
import pandas as pd

df_knn = students.copy()
df_knn = df_knn.rename(columns={'Msrks': 'Marks', 'Attendence': 'Attendance'})
knn_imputer = KNNImputer(n_neighbors=2)
numeric_cols = ['Marks', 'Attendance']
df_knn[numeric_cols] = knn_imputer.fit_transform(df_knn[numeric_cols])
print(df_knn)
```

```python

clean = students.copy()
clean = clean.rename(columns={'Msrks': 'Marks', 'Attendence': 'Attendance'})
clean['Marks'] = clean['Marks'].fillna(clean['Marks'].mean())
clean['Attendance'] = clean['Attendance'].fillna(clean['Attendance'].median())
clean['Name'] = clean['Name'].fillna('Unknown')

print(clean)
```
