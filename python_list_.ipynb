# List
## list is ordered collection of items stored in[]


```python
#creat a list
marks = [67,2,4,242,312,55,2222]
print(marks)
```

```python
student_info=[101,"rahul","cse"]
print(student_info)
```

```python
# access the element using index
name=["aneesh","chirag","sneha","ram","devi"]
print(name)
print(name[2])
print(name[3])
print(name[1])
print(name[-1])
print(name[-3])
```

```python
# example for mutable in the list
cgpa=[8.9,9.4,4.6,6.8]
print[cgpa]
cgpa[2]=
```

```python
# add an item in the list
name=["sonam","paras","lalit","chirag"]
name.append("deepika")
print(name)
```

```python
# insert in particular index
student_name=["arrav","kapil","hariram"]
student_name.insert(2,"ram")
print(name)
```

```python
# extend function
sec_a=["chirag","ram","raghav","dev"]
sec_b=["sonam","paras","lalit"]
sec_a.extend(sec_b)
print(sec_a)
print(sec_b)
```

```python
# remove an item
sec_a_students=["chirag","ram","raghav","dev"]
sec_a_students.remove("raghav")
print(sec_a_students)
del sec_a_students[1]
print(sec_a_students)
```

```python
# list slicing
subjets=["python",
         "statics",
         "sql",
         "java","os",
         "machine learning"]
print(subjets[1:4])
print(subjets[:3])
print(subjets[4:])
print(subjets[::2])

```

```python
# sorting in ascending and descending
marks=[78,32,32,45,67,88,96]
marks.sort()
print(marks)
marks.sort(reverse=True)
print(marks)
```

```python
# nested lists
marks=[
    [91,42,64],
    [69,"oi",21],
    [65,74,82]
    ]
print(marks)
print(marks[1])
print(marks[1][1])
```
